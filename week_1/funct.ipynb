```python
def mix_and_cook():
    print('biriyani preparetion')
    print('mixing the ingredients')
    print('cooking the rice one side')
    print('another side cooking the biriyani')
    print('after it together')

def make_biriyani(ingredient):    #parameter
    mix_and_cook()
    biriyani = 'a {} biriyani'.format(ingredient)
    return biriyani


def make_idli():
    mix_and_cook()
    idli = 'a testy idli'
    return idli


def make_omelette(*ingredients):
    mix_and_cook()
    omelette = 'a fancy omelette with {} ingredients'.format(len(ingredients))
    return omelette
```

```python
make_biriyani('chicken')    #argument
```

```python
make_biriyani('mutton')    #argument
```

```python
print(make_omelette('sausage','onion','pepper','spicach','mushroom','tomato','cheese'))
```
